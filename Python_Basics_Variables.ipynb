# Variables in Python

* A variables is a name that refers to a vale stored in memory.
* You can assign values using the '=' operator. Python is dynamically typed.

**Rules for naming variables**

1. Must start with a letter or underscore_
2. Cannot start with a number
3. Can contain letters, digits and underscore
4. Case sensitive
5. Should not use Python Keywords
